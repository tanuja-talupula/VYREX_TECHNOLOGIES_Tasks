```python
# Define a DNA sequence
dna = "ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA"

# Count nucleotides using a dictionary
counts = {}
for i in dna:
    if i not in counts:
        counts[i] = 1
    else:
        counts[i] += 1
print(counts)
```

```python
# Calculate GC content
gc_content = (counts['G'] + counts['C']) / len(dna) * 100
print(f"GC Content: {gc_content:.2f}%")
```

```python
# Plot a bar chart of counts using matplotlib
import matplotlib.pyplot as plt

plt.bar(counts.keys(), counts.values())
plt.xlabel("Nucleotide")
plt.ylabel("Count")
plt.title("DNA Nucleotide Frequency")
plt.show()
```

# **DNA Nucleotide Frequency and GC Content Analysis**

#### **Objective**

To calculate the frequency of A, T, G, and C nucleotides in a DNA sequence,
determine the GC content, and visualize nucleotide frequencies using a bar chart.

#### **Method**

A Python dictionary is used to count the occurrences of each nucleotide.
The GC content is calculated using the formula:

$$
GC\% = \frac{G+C}{A+T+G+C}\times100
$$

The nucleotide frequencies are then visualized using Matplotlib.

#### **Result**

The calculated GC content is: **50.00%**

#### **Interpretation**

The GC content of **50.00%** means that half of the nucleotides in the DNA
sequence are **Guanine (G) or Cytosine (C)**, while the other half are
**Adenine (A) or Thymine (T)**.

This indicates a relatively balanced **GC and AT composition** in the DNA
sequence. GC content is an important sequence characteristic commonly used
in bioinformatics and genomic analysis.

The bar chart provides a visual representation of the frequency of each
nucleotide and helps identify the overall nucleotide composition of the
sequence.
